## 1. Load the two batches
Run from this lab folder. Only two derived features are used; the saved SMS text and IDs stay out of Evidently.

```python
from pathlib import Path
from monitoring import load_features, summarize, make_report, export_report

ROOT = Path.cwd()
reference = load_features(ROOT / "data/reference_sms.csv")
current = load_features(ROOT / "data/current_sms.csv")
reference.head()
```

## 2. Inspect the features
`message_length` counts stored characters, including spaces. `has_url` is `yes` when the lowercased text contains `http://`, `https://` or `www.`. It is an approximate signal, not a spam label. These are the same rules as Lab 2; monitoring.py contains the short supplied implementation.

```python
import pandas as pd
pd.DataFrame({"Reference": summarize(reference), "Current": summarize(current)})
```

## 3. Generate the report
The supplied code explicitly declares length as numerical and URL presence as categorical. The tested Evidently API receives **current first, reference second**. No raw-text drift classifier is run.

```python
result = make_report(reference, current)
report_path = export_report(result, reference, current, ROOT / "output")
print(report_path)
```
